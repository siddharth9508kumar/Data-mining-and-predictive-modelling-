```python
# to implement the frequent mining using apriori alogrithem

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

# create sample dataset
data = [
    ["Bread", "Milk"],
    ["Bread", "Diaper", "Beer", "Eggs"],
    ["Milk", "Diaper", "Beer", "Cola"],
    ["Bread", "Milk", "Diaper", "Beer"],
    ["Bread", "Milk", "Diaper", "Cola"]
]

# perform encoding
te = TransactionEncoder()
te_ary = te.fit(data).transform(data)
df = pd.DataFrame(te_ary,columns=te.columns_)

# to print
print(" the encoded data is : ")
print(df.head())
print("\n")


# 3. Apply Apriori Algorithm to find frequent itemsets
# min_support = 0.6 means the itemset must appear in at least 60% (3 out of 5) of the transactions
min_support_threshold = 0.6
frequent_itemsets = apriori(
    df, min_support=min_support_threshold, use_colnames=True
)

print(f"--- Frequent Itemsets (Min Support = {min_support_threshold}) ---")
print(frequent_itemsets)
print("\n" + "=" * 40 + "\n")

# 4. Generate Association Rules from the frequent itemsets
# min_threshold = 0.7 means we filter out rules with confidence less than 70%
rules = association_rules(
    frequent_itemsets, metric="confidence", min_threshold=0.7
)

# Sorting rules by lift to see the strongest associations first
rules = rules.sort_values(by="lift", ascending=False)

print("--- Generated Association Rules (Min Confidence = 0.7) ---")
print(
    rules[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift",
        ]
    ]
)


```

```python
import warnings
warnings.filterwarnings('ignore')
# Q2 : implement fp-growth alogorithem

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth, association_rules

# 1. Define the sample dataset
data = [
    ["Bread", "Milk"],
    ["Bread", "Diaper", "Beer", "Eggs"],
    ["Milk", "Diaper", "Beer", "Cola"],
    ["Bread", "Milk", "Diaper", "Beer"],
    ["Bread", "Milk", "Diaper", "Cola"]
]

# 2. Format the dataset into a One-Hot Encoded DataFrame
te = TransactionEncoder()
te_ary = te.fit(data).transform(data)
df = pd.DataFrame(te_ary, columns=te.columns_)

print("--- One-Hot Encoded DataFrame ---")
print(df)
print("\n")

# 3. Apply FP-Growth Algorithm
# min_support = 3 out of 5 transactions = 0.6
frequent_itemsets = fpgrowth(df, min_support=0.6, use_colnames=True)

print("--- Frequent Itemsets (Support >= 0.6) ---")
print("\n")
print(frequent_itemsets)
print("\n")

# 4. Generate Association Rules
# Extracts actionable insights using a minimum confidence threshold of 70%
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

print("--- Generated Association Rules ---")
print("\n")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])

```
