# PERFORM NORMALIZATION ON data =[10,20,30,40,50]


```python
import math
import numpy as np
# Create a sample dataset
data =[10,20,30,40,50]

# to perform min-max normalization
min_value = min(data)
max_value = max(data)

normalized_data = [(x - min_value) / (max_value - min_value) for x in data]

print("Original Data:", data)
print("Normalized Data by min-max nor..:", normalized_data)

# to perform z-score normalization
mean_value = np.mean(data)
std_dev = np.std(data)

normalized_data_z = np.array([(x - mean_value) / std_dev for x in data])
print("Original Data:", data)
print("Normalized Data by z-score nor..:", normalized_data_z)

# to perform decimal scaling normalization
# Find the maximum absolute value in the data
max_abs_value = max(abs(x) for x in data)

# Determine 'j' such that max_abs_value / (10**j) < 1
# This means 10**j > max_abs_value
if max_abs_value == 0:
    j = 0
else:
    j = math.ceil(math.log10(max_abs_value))

normalized_data_decimal = [x / (10**j) for x in data]

print("Original Data:", data)
print(f"Normalized Data by decimal scaling nor.. (j={j}):", normalized_data_decimal)
```

# TO PERFORM CORRELATION COEFFICIENT RELATION AND CO-VARIANCE ANALYSIS - chi square test


```python
import numpy as np
import pandas as pd
import scipy.stats as stats


# let a dataset
data={
    'hour':[2,4,6,8,10],
    'marks':[40,50,60,70,80]
}
df=pd.DataFrame(data)
print(df)


# Create a contingency table (Cross-tabulation)
contingency_table = pd.crosstab(df['hour'], df['marks'])

# Perform the chi-square test
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
print("Chi-square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)


# conclude the answer
if p_value < 0.05:
    print("Reject the null hypothesis. There is a significant relationship between 'hour' and 'marks'.")
else:
    print("Fail to reject the null hypothesis. There is no significant relationship between 'hour' and 'marks'")
```
