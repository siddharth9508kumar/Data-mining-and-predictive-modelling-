```python
 # A program to find the mean ,medin,mode and standard daviartion and variance ..

import numpy as np
from scipy import stats

x1=[2,3,4,5,6,8,10,11,14,12,16,18,10]
mean=np.mean(x1)
print("mean of the data is",mean)
median=np.median(x1)
print("median of the data is ",median)

# For mode, we need to use scipy.stats.mode
mode_result = stats.mode(x1, keepdims=True)
mode = mode_result.mode[0] if len(mode_result.mode) > 0 else 'No unique mode'
print("mode of the data is",mode)

#for std and variance
std_dev=np.std(x1)
print("Std daviation is",std_dev)
variance=np.var(x1)
print("variance is",variance)


# now without function


mean=sum(x1)/len(x1)
print("mean of the data is",mean)

mid = len(x1) // 2
median = (x1[mid - 1] + x1[mid]) / 2
print("median of the data is",median)



# Count the frequency of each number
counts = {}
for num in x1:
    if num in counts:
        counts[num] += 1
    else:
        counts[num] = 1

# Find the number with the highest frequency

mode = x1[0]
max_count = 0

for num, count in counts.items():
    if count > max_count:
        max_count = count
        mode = num

print("Mode:", mode)

```
