```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# take a sample data
dataset= {
    'x': [1,2,3,4,5],
    'y': [2,4,6,8,10]
}

data=pd.DataFrame(dataset)
data.head()

# for data visualization
plt.plot(data['x'],data['y'])
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

#scatter plot
plt.scatter(data['x'],data['y'])
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

#bar plot
plt.bar(data['x'],data['y'])
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

# pie chart
plt.pie(data['y'],labels=data['x'])
plt.title('Sample Plot')
plt.show()

#histograms
plt.hist(data['y'])
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

#box plot
plt.boxplot(data['y'])
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

```

```python
# some seaborn functions are
sns.displot(data['y'])
sns.boxplot(data['y'])
sns.histplot(data['y'])
sns.barplot(x=data['x'],y=data['y'], data=data)
sns.scatterplot(x=data['x'],y=data['y'], data=data)
sns.heatmap(data)
sns.pairplot(data)
```
